```python
import osmium
import pandas as pd
```

```python
class OSMHandler(osmium.SimpleHandler):
    """Process OSM data."""

    def __init__(self):
        """Initialise Handler."""
        super().__init__()
        self.node_count = 0
        self.way_count = 0
        self.relation_count = 0
        self.restaurants = []

    def node(self, n):
        """Process a node element from the OSM data."""
        tags = dict(n.tags)

        if "restaurant" in tags.get("amenity", ""):
            self.restaurants.append(tags)

    def way(self, w):
        """Process a way element from the OSM data."""
        self.way_count += 1

    def relation(self, r):
        """Process a relation element from the OSM data."""
        self.relation_count += 1
```

```python
handler = OSMHandler()
handler.apply_file("../data/raw/karlsruhe-regbez-latest.osm.pbf")
```

```python
print(f"Restaurants: {pd.DataFrame(handler.restaurants).head(5)}")
```
